## Question 1 — Simple Interest Calculator (Default Arguments)

Write a function `calculate_simple_interest(principal, rate=5, time=1)` that calculates and
returns the simple interest using the formula:

```
interest = (principal * rate * time) / 100
```

- `rate` should default to `5` (percent) if not provided.
- `time` should default to `1` (year) if not provided.

**Sample calls and expected output:**
```python
calculate_simple_interest(1000, 10, 2)   # -> 200.0
calculate_simple_interest(1000)          # -> 50.0   (uses default rate=5, time=1)
calculate_simple_interest(2000, time=3)  # -> 300.0  (uses default rate=5)
```

````python

def calculate_simple_interest(principal, rate=5, time=1):
    # principal: the initial amount of money
    # rate: annual interest rate in percent, defaults to 5 if not provided
    # time: time period in years, defaults to 1 if not provided

    # Simple interest formula: (Principal * Rate * Time) / 100
    interest = (principal * rate * time) / 100

    return interest  # return the calculated interest value


# Call 1: principal=1000, rate=10, time=2 (all args given explicitly)
# Interest = (1000 * 10 * 2) / 100 = 200.0
print(calculate_simple_interest(1000, 10, 2))

# Call 2: principal=1000, rate and time use defaults (5 and 1)
# Interest = (1000 * 5 * 1) / 100 = 50.0
print(calculate_simple_interest(1000))

# Call 3: principal=2000, rate uses default (5), time=3 passed as a keyword argument
# Interest = (2000 * 5 * 3) / 100 = 300.0
print(calculate_simple_interest(2000, time=3))
````

## Question 2 — Class Average Using `*args`

Write a function `class_average(*scores)` that accepts **any number of scores** and returns
their average, rounded to 2 decimal places.

- If no scores are passed at all, the function should return `0`.

**Sample calls and expected output:**
```python
class_average(80, 90, 70)         # -> 80.0
class_average(55, 60, 65, 70, 75) # -> 65.0
class_average()                   # -> 0
```

````python

def class_average(*scores):
    # *scores collects any number of arguments passed in as a tuple
    # e.g. class_average(80, 90, 70) -> scores = (80, 90, 70)
    try:
        # Try to calculate the average: sum of all scores divided by how many there are
        return sum(scores) / len(scores)

    except ZeroDivisionError:
        # If no scores were passed, len(scores) is 0, which causes division by zero
        # In that case, just return 0 instead of crashing
        return 0

# --- test your function below ---

# Test 1: average of 80, 90, 70 -> (80+90+70)/3 = 80.0
print(class_average(80, 90, 70))

# Test 2: average of 55, 60, 65, 70, 75 -> (55+60+65+70+75)/5 = 65.0
print(class_average(55, 60, 65, 70, 75))

# Test 3: no scores given -> triggers ZeroDivisionError -> caught and returns 0
print(class_average())
````

## Question 3 — Analyze Numbers (Multiple Return Values + Built-ins)

Write a function `analyze_numbers(numbers)` that takes a list of numbers and returns **four
values** in this order:
1. the smallest number
2. the largest number
3. the sum of all numbers
4. the numbers sorted in descending order (as a list)

Use the built-in functions `min()`, `max()`, `sum()`, and `sorted()` — do not calculate these
manually with loops.

**Sample call and expected output:**
```python
smallest, largest, total, desc = analyze_numbers([4, 9, 1, 7, 3])
print(smallest)   # -> 1
print(largest)    # -> 9
print(total)      # -> 24
print(desc)       # -> [9, 7, 4, 3, 1]
```

````python
# Write your code here
def analyze_numbers(numbers):
    smallest = min(numbers)              # get the smallest number in the list
    largest = max(numbers)               # get the largest number in the list
    total = sum(numbers)                 # add up all numbers in the list
    desc = sorted(numbers, reverse=True) # sort numbers from largest to smallest
    return smallest, largest, total, desc  # return all 4 values as a tuple

# --- test your function below ---

# unpack the returned tuple into 4 separate variables
smallest, largest, total, desc = analyze_numbers([4, 9, 1, 7, 3])

print(smallest)   # 1
print(largest)    # 9
print(total)      # 24
print(desc)       # [9, 7, 4, 3, 1]
````

## Question 4 — Shared Booking Counter (Scope & the `global` keyword)

A small ticket-booking system needs to keep track of how many seats have been booked so far,
using a **global variable**.

1. Create a global variable `total_seats_booked` initialized to `0`.
2. Write a function `book_seats(n)` that:
   - adds `n` to `total_seats_booked` using the `global` keyword
   - prints a message like: `"Booked 3 seat(s). Total booked so far: 3"`
3. Write a second function `reset_bookings()` that resets `total_seats_booked` back to `0`
   (this also needs the `global` keyword).

**Sample calls and expected behaviour:**
```python
book_seats(3)     # Booked 3 seat(s). Total booked so far: 3
book_seats(5)     # Booked 5 seat(s). Total booked so far: 8
reset_bookings()
book_seats(2)     # Booked 2 seat(s). Total booked so far: 2
```

````python
# Initialize the global variable before any function uses it.
# This must exist first because book_seats() reads its value (total_seats_booked + n)
# before reassigning it — without this line, that read would fail with NameError.
total_seats_booked = 0

def book_seats(n):
    # Tell Python to use the global total_seats_booked, not create a new local one
    global total_seats_booked
    
    # Read the current value, add n seats, then store the new total back globally
    total_seats_booked = total_seats_booked + n
    
    # Return the updated total after booking
    return total_seats_booked


def reset_bookings():
    #use the global total_seats_booked, not create a new local one
    global total_seats_booked
    
    # Reset the total back to 0 (pure assignment, no read needed here)
    total_seats_booked = 0
    
    # Return the reset value (0) to confirm the reset worked
    return total_seats_booked


# --- test your functions below ---

print(book_seats(3))      # 0 + 3 = 3
print(book_seats(5))      # 3 + 5 = 8
print(reset_bookings())   # reset back to 0
print(book_seats(2))      # 0 + 2 = 2
````
